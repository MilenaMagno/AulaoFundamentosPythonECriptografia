```python
idade = 15 # variável

if (idade < 10):          # condicional "se/if" verifica a condição
  print("Você é uma criança")     # se for verdadeira, execura a instrução
elif (idade < 18):
  print("Você é um adolescente")
elif (idade < 60):
  print("você é adulto")
else:                           # "senão/else" caso contrário (nenhuma condição a cima era verdadeira)
  print("Você é idoso")         # não tem verificação, já faz a instrução
```
